## 要素の取り出しと型変換

* **インデックス**：リスト・タプル・文字列の要素は`[番号]`で取り出す。番号は**0から始まる**（1番目の要素は`[0]`）。`[-1]`は最後の要素。
* **スライス**：`[開始:終了]`で範囲を取り出す。終了の番号は含まない（`e[1:3]`は`e[1]`と`e[2]`）。
* リストの要素は`e[1] = 20`のように変更できるが、タプルの要素は変更できない。
* **アンパック**：`width, height = (1920, 1080)`のように、タプルの要素を複数の変数に一度に代入できる。
* **辞書**：`h["physics"]`のようにキーを指定して値を取り出す。
* **型変換**：`int("123")`（文字列→整数）、`float("3.5")`、`str(10)`（数値→文字列）
* **f文字列**：`f"..."`の中に`{変数}`と書くと、変数の値が埋め込まれる。

````python
e = [1, 2, 3, 4, 5]
print(e[0], e[1], e[-1]) # 1番目、2番目、最後の要素
print(e[1:3])            # 2番目から3番目まで（e[3] は含まない）
e[1] = 20                # 2番目の要素を変更
print(e, len(e), sum(e)) # 要素の数と合計

width, height = (1920, 1080) # アンパック
print(width, height)

h = {"biology": 70, "chemistry": 68, "physics": 75}
print(h["physics"])      # 辞書はキーで取り出す

n = int("123")           # 文字列を整数に変換
name = "taro"
print(f"Hello, {name}! n+1={n+1}") # f文字列
````

## リスト内包表記*

python独特の記法。短く書けるのでよく使われる。

```python
[式 for 変数 in 繰り返し要素]
```
forループを1行に短縮したような形
```python
リスト名=[]
for 変数 in 繰り返し要素:
    リスト名.append(式)
```

````python
# 例１
numbers = [1, 2, 3, 4, 5]
doubled = [x*2 for x in numbers]
print(doubled)

# for 文で書くと
doubled = [] # 空のリストを作る
for x in numbers:
    doubled.append(x*2) # 要素を追加する
print(doubled)
````

````python
# 例2
# 3項演算子（条件式）と組み合わせる
numbers = [1, 2, 3, 4, 5]
# 偶数ならそのまま、奇数なら0にする
result = [x if x % 2 == 0 else 0 for x in numbers]
print(result)  # 出力: [0, 2, 0, 4, 0]

# 同じことをfor文で書くと
result = []
for x in numbers:
    if x % 2 == 0:
        result.append(x)
    else:
        result.append(0)
print(result)
# 条件が3つ以上になる場合は、for文で書いた方が読みやすい。
````

## 3項演算子*
```python
真のときの値 if 条件 else 偽のときの値
```
リスト内包表記、3項演算子は簡潔に書けるが、慣れるまでは`for`文や`if`文が使えればよい。 

````python
# 例
a = 10
b = 20
max_value = a if a > b else b # a > b のときは a, それ以外のときは b を返す
print(a, b, max_value)

# if 文で書くと
a = 20
b = 10
if a > b:
    max_value = a
else:
    max_value = b
print(a, b, max_value)
# 4行分を1行に圧縮
````

## 数値の丸め*

Pythonの`round()`は**高校までの四捨五入と結果が違うことがある**ので注意。

| 関数 | 動作 | 2.5 | 3.5 | -2.5 |
|---|---|---|---|---|
| 高校の四捨五入 | 5以上は切り上げ | 3 | 4 | -3 |
| `round()` | 偶数への丸め | **2** | 4 | **-2** |
| `numpy.round()` | 偶数への丸め | **2.0** | 4.0 | **-2.0** |
| `math.floor()` | 切り捨て（小さい方の整数） | 2 | 3 | -3 |
| `math.ceil()` | 切り上げ（大きい方の整数） | 3 | 4 | -2 |

### なぜ違うのか
1. **偶数への丸め（Round Half to Even）**：ちょうど真ん中（…5）のときだけ、近い**偶数**の方に丸める。
   いつも切り上げると、大量のデータを丸めて合計したときに少しずつ大きい方へずれていく。偶数への丸めは切り上げと切り捨てが半々になるので、この偏りが生じない。統計や科学計算の標準的な方法（JIS Z 8401 でも採用）。
2. **小数は2進数で正確に表せない**：`2.675` はコンピュータの中では `2.67499999...` なので、`round(2.675, 2)` は 2.68 ではなく **2.67** になる（「真ん中」ではないので、偶数への丸め以前の問題）。

### 他のソフトでは
Excelの`ROUND()`、Cの`round()`、Fortranの`NINT()`は高校と同じ四捨五入（-2.5は-3）。
一方、Excel VBAの`Round()`、Rの`round()`はPythonと同じ偶数への丸め。
**ExcelとPythonで結果が1ずれたら、丸め方の違いを疑う。**

### どうすればよいか
* データ解析・機械学習では`round()`のままでよい（差はごくわずかで、むしろ偏りがない）。
* 成績や金額など、**高校の四捨五入どおりの結果が必要なとき**は`decimal`ライブラリを使う（下の`shishagonyu()`関数）。

````python
# round() の結果を予想してから実行してみよう
print(round(0.5), round(1.5), round(2.5), round(3.5)) # 偶数への丸め
print(round(2.675, 2)) # 小数第2位に丸める。2進数の誤差で 2.67 になる
````

````python
# 高校の四捨五入どおりに丸める関数
from decimal import Decimal, ROUND_HALF_UP

def shishagonyu(x, ndigits=0):
    # str(x) で「見たままの数字」にしてから丸めるので 2進数の誤差の影響を受けない
    q = Decimal(1).scaleb(-ndigits) # ndigits=0 なら 1, ndigits=2 なら 0.01
    r = Decimal(str(x)).quantize(q, rounding=ROUND_HALF_UP)
    return int(r) if ndigits == 0 else float(r)

print(shishagonyu(2.5), shishagonyu(3.5), shishagonyu(-2.5))
print(shishagonyu(2.675, 2))
````

## 例：測定データファイルの仕分け

顕微鏡やプレートリーダーなどの測定装置からは、毎日多数のファイルが出力される。たとえば次のような状況を考える。

* ファイル名はサンプルIDで始まる（例：`S001_field1.tif`）
* 1つのサンプルに複数のファイル（視野、チャンネルなど）があることもある
* サンプルIDと実験群・解析用番号の対応表をExcelファイルで管理している

毎日、出力されたファイルを実験群ごとのフォルダに分け、ファイル名を解析用番号に揃えたい。100個以上のファイルを手作業で処理すると時間がかかり、間違いも起きやすい。コードにしておけば、対応表を更新するだけで毎日同じ処理を繰り返せる。
同じ方法は、学籍番号で始まるレポートファイルを採点グループ別に仕分けるような事務作業にも使える。

ここでは、各ファイルを**どのフォルダに、どの名前で**置くかを決める部分を作る。使うのは、ここまでに出てきた**辞書、for ループ、if 文、f文字列、スライス**が中心である。

### データの準備
練習用のファイル（中身は空）を119個まとめたZIPファイル raw.zip を[ここ](https://github.com/kyorin-phys/MLIntro2026/blob/main/1/data/raw.zip)からダウンロードして、「ファイル」からアップロードする。ファイルを1つずつアップロードする必要はない。

````python
# ! で始まる行はPythonではなく、Colabのコマンド（unzip でZIPファイルを展開する）
!unzip -oq raw.zip

import os
files = sorted(os.listdir("raw"))  # raw フォルダ内のファイル名のリスト
print(len(files), "個のファイル")
print(files[:10])  # 最初の10個
````

````python
# 対応表（ダミーデータ）　サンプルID: (実験群, 解析用番号)
table = {
    "S001": ("treated", 12), "S002": ("treated", 5), "S003": ("treated", 11), "S004": ("treated", 33),
    "S005": ("treated", 8), "S006": ("treated", 15), "S007": ("control", 35), "S008": ("treated", 7),
    "S009": ("control", 38), "S010": ("control", 36), "S011": ("control", 31), "S012": ("control", 3),
    "S013": ("control", 22), "S014": ("control", 27), "S015": ("treated", 10), "S016": ("control", 1),
    "S017": ("control", 21), "S018": ("control", 23), "S019": ("treated", 32), "S020": ("control", 9),
    "S021": ("control", 25), "S022": ("treated", 13), "S023": ("treated", 34), "S024": ("treated", 39),
    "S025": ("treated", 17), "S026": ("control", 14), "S027": ("control", 4), "S028": ("control", 16),
    "S029": ("control", 18), "S030": ("control", 19), "S031": ("treated", 29), "S032": ("treated", 26),
    "S033": ("treated", 6), "S034": ("treated", 28), "S035": ("control", 30), "S036": ("control", 40),
    "S037": ("treated", 24), "S038": ("treated", 2), "S039": ("treated", 20), "S040": ("control", 37),
}
````

````python
count = 0
for name in files:
    sid = name[:4]  # 先頭の4文字がサンプルID
    if sid not in table:
        print("対応表にない:", name)
        continue  # このファイルは飛ばして次へ
    group, num = table[sid]
    new_name = f"{num:03d}{name[4:]}"  # サンプルIDを3桁の解析用番号に置き換える
    print(name, "->", f"{group}/{new_name}")
    count += 1
print(count, "個のファイルを処理した")
````

結果を見ると、次のことも確認できる。
* 複数のファイルがあるサンプルは、すべてのファイルが処理されている
* 対応表にないIDのファイル（`memo.txt`のようなファイルも含む）は、処理されずに表示される
* 対応表にあってもファイルがないサンプルは何も起きない。測定漏れの確認は別に行う必要がある（練習：ファイルがないサンプルIDを表示するコードを書いてみよう）

実際にファイルをコピーするには、`shutil`などのライブラリの関数を使う（ここでは扱わない）。
実際にファイルを操作する前に、このように**どのファイルがどこに行くかを表示して確認する**とよい。ファイル操作は失敗すると元に戻すのが大変なので、移動ではなくコピーにして元のファイルを残しておく。
