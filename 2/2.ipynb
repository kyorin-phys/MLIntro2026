## numpy 配列

`numpy.array()`

リストの場合は要素ごとに演算が必要だが、numpy配列ではまとめて演算できる。

例：すべての値を2倍する

### リストの場合

````python
# 2倍するつもりが、リストをつなぐ結果に
a = [1, 2, 3, 4, 5]
print(a * 2)
print(a + a)
````

````python
# 全ての要素を2倍する
a = [1, 2, 3, 4, 5]
a = [x*2 for x in a]  #リスト内包表記
print(a)

# for文は遅いのでなるべく避けた方がよい
a = [1, 2, 3, 4, 5]
for i in range(len(a)):
    a[i] *= 2  # i番目の要素を2倍
print(a)
````

### numpyの場合

````python
# ベクトルの定数倍と同様にすべての要素が2倍になる
b = np.array([1, 2, 3, 4, 5])
print(b)
print(b * 2)
print(type(b))
````

````python
# 配列のインデックスの代わりに条件式でフィルターできる
b[b < 3.5], b[b % 2 == 0] # bの要素で値が3.5より小さい、bの要素で偶数だけを選ぶ
````

````python
# 積は同じインデックスの要素同士を掛ける
a = [1, 2, 3, 4, 5]
print(a * b) # これは可能
print(b * b)
c = np.array([1,2,3])
#print(b*c) # 長さの違う配列の積は計算できないのでエラー
````

## 行列 多次元配列

* 1次元の配列: ベクトル(vector)
* 2次元の配列: 行列(matrix)
* 3次元以上の配列: テンソル(tensor)

例:モノクロ画像は2次元のデータからなる。
カラー画像は色情報がRGBの3チャンネルを持つ2次元データを組み合わせたもの。

CTなどの画像データは3次元テンソル (x,y,z)座標に対するCT値の組。
スライスで2次元の画像に切り出したものを画像として見る。

````python
# matplotlib.imshow で画像を表示させる
import matplotlib.pyplot as plt

# 2x2のモノクロ画像（グレースケール）を作成
monochrome_image = [[0, 127], [255, 0]]
#monochrome_image = np.array([[0, 127], [255, 0]])
# 0　真っ黒、255　真っ白
# モノクロ画像を表示
plt.imshow(monochrome_image, cmap='gray')
plt.axis('off')
plt.show()
````

````python
# 2x2のカラー画像（RGB）を作成
color_image = [[[255, 0, 0], [0, 255, 0]],
               [[0, 0, 255], [255, 255, 0]]]
#color_image = np.array([[[255, 0, 0], [0, 255, 0]],
#               [[0, 0, 255], [255, 255, 0]]])
# [R,G,B]: [255,0,0] は赤、[0,255,0]は緑、[0,0,255]は青、[255,255,0]は黄
# [[赤,緑],[青,黄]] のように色情報を要素に持つ2x2の画像になっている　(2x2x3：3次元テンソル）
# カラー画像を表示
plt.imshow(color_image)
plt.axis('off')
plt.show()
````

````python
# グレースケール
gradient_image = np.linspace(0, 255, 10, dtype=np.uint8).reshape(1, 10) # ベクトルを1x10の行列に変換
# linspace: linear space 区間を等間隔に分割した等差数列を返す
# linspace(start, stop, num)
print(gradient_image)
plt.imshow(gradient_image, 'gray')
plt.axis('off')
plt.show()
````

## サンプル画像と3次元データ*

scikit-image (`skimage`) には、医療・生物の画像を含むサンプル画像が入っている。`data.関数名()` でnumpy配列として取り出せる（一部は初回にダウンロードするので時間がかかる）。

````python
# サンプル画像の一覧を表示する
from skimage import data
import math

images = {}
for name in data.__all__:
    if name in ('download_all', 'binary_blobs', 'lbp_frontal_face_cascade_filename', 'data_dir', 'file_hash'):
        continue  # 画像を返さない関数は飛ばす
    try:
        img = getattr(data, name)()
    except Exception as e:
        print(f'{name}: 読み込めない ({type(e).__name__})')
        continue
    if isinstance(img, np.ndarray) and (img.ndim == 2 or (img.ndim == 3 and img.shape[-1] in (3, 4))):
        images[name] = img
    else:
        shape = img.shape if isinstance(img, np.ndarray) else type(img).__name__
        print(f'{name}: 2次元画像ではない {shape}')  # 3次元データ・複数枚の画像など

cols = 6
rows = math.ceil(len(images) / cols)
fig, axes = plt.subplots(rows, cols, figsize=(cols * 2.5, rows * 2.8))
for ax in axes.flat:
    ax.axis('off')
for ax, (name, img) in zip(axes.flat, images.items()):
    if img.dtype != np.uint8:          # 16bitの蛍光画像などは0〜1に正規化
        img = img[..., :3] / img.max()
    ax.imshow(img, cmap='gray' if img.ndim == 2 else None)
    ax.set_title(f'{name}\n{img.shape}', fontsize=8)
plt.tight_layout()
plt.show()
````

`data.cells3d()` は細胞の核と細胞膜を蛍光顕微鏡で深さを変えながら60枚撮影した3次元データ。
形は (深さz, チャンネル, 縦, 横) = (60, 2, 256, 256) の4次元配列。

CTやMRIも同じように3次元データであり、スライス（断面）を切り出して2次元の画像として見る。

````python
c = data.cells3d()[:, 1]   # チャンネル1 = 核の染色
print(c.shape)             # (60, 256, 256)

fig, axes = plt.subplots(1, 5, figsize=(18, 4))
for ax, z in zip(axes[:3], [10, 30, 50]):
    ax.imshow(c[z], cmap='gray')       # 深さzの断面
    ax.set_title(f'z={z}')
axes[3].imshow(c[:, 128, :], cmap='gray', aspect=0.29/0.26)  # 横から見た断面（深さ方向の画素間隔0.29µm、平面方向0.26µm）
axes[3].set_title('c[:, 128, :]')
axes[4].imshow(c.max(axis=0), cmap='gray')  # 深さ方向の最大値投影
axes[4].set_title('c.max(axis=0)')
for ax in axes:
    ax.axis('off')
plt.show()
````
