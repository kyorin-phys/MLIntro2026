## numpy 配列

`numpy.array()`

リストの場合は要素ごとに演算が必要だが、numpy配列ではまとめて演算できる。

例：すべての値を2倍する

### リストの場合

````python
# 2倍するつもりが、リストをつなぐ結果に
a = [1, 2, 3, 4, 5]
print(a * 2)
print(a + a)
````

````python
# 全ての要素を2倍する
a = [1, 2, 3, 4, 5]
a = [x*2 for x in a]  #リスト内包表記
print(a)

# for文は遅いのでなるべく避けた方がよい
a = [1, 2, 3, 4, 5]
for i in range(len(a)):
    a[i] *= 2  # i番目の要素を2倍
print(a)
````

### numpyの場合

````python
# ベクトルの定数倍と同様にすべての要素が2倍になる
b = np.array([1, 2, 3, 4, 5])
print(b)
print(b * 2)
print(type(b))
````

````python
# 配列のインデックスの代わりに条件式でフィルターできる
b[b < 3.5], b[b % 2 == 0] # bの要素で値が3.5より小さい、bの要素で偶数だけを選ぶ
````

````python
# 全ての要素に2を足す
a = [1, 2, 3, 4, 5]
#print(a + 2) # リストと数値は足せない
#print([1,2,3,4,5]+[2,2,2,2,2])
#for i in range(5):
#    a[i] += 2
#print(a)
#a = [x+2 for x in a] # リスト内包表記   
#print(a)
# broadcast 2*np.oneslike(b) 
print(b + 2)
print(np.ones_like(b))
#print(b + 2*np.ones_like(b))
````

````python
# 積は同じインデックスの要素同士を掛ける
a = [1, 2, 3, 4, 5]
print(a * b) # これは可能
print(b * b)
c = np.array([1,2,3])
#print(b*c) # 長さの違う配列の積は計算できないのでエラー
````

````python
## ベクトルの内積
a = np.array([1,2])
b = np.array([2,-1])
print(np.dot(a, b)) # 1*2 + 2*(-1)
````

## numpy の関数
import numpy as np

* np.linspace(a,b,n) 等差数列を作る（要素の数を指定する）
* np.arange(a,b,dx) 等差数列を作る（幅を指定する）
* np.reshape(A, (h, w)) numpy配列Aの形状を縦h、横wに変更（変更前と変更後で要素数は一致）

**配列の形状を調べたり、形を変更することは機械学習のコードを動かすうえで必ず必要になる。**

````python
A = np.linspace(0, 10, 10) # 0から10までの区間を(10-1)等分して要素10個のnumpy配列を作る
print(A.shape)
print(A)
print("-" * 80)
B  = np.reshape(A, (2,5)) # 2行5列に変換
print(B)
print("-" * 80)
C = np.reshape(A, (5,2)) # 5行2列に変換
print(C)
````

## 行列 多次元配列

* 1次元の配列: ベクトル(vector)
* 2次元の配列: 行列(matrix)
* 3次元以上の配列: テンソル(tensor)

例:モノクロ画像は2次元のデータからなる。
カラー画像は色情報がRGBの3チャンネルを持つ2次元データを組み合わせたもの。

CTなどの画像データは3次元テンソル (x,y,z)座標に対するCT値の組。
スライスで2次元の画像に切り出したものを画像として見る。

````python
# matplotlib.imshow で画像を表示させる
import matplotlib.pyplot as plt

# 2x2のモノクロ画像（グレースケール）を作成
monochrome_image = [[0, 127], [255, 0]]
#monochrome_image = np.array([[0, 127], [255, 0]])
# 0　真っ黒、255　真っ白
# モノクロ画像を表示
plt.imshow(monochrome_image, cmap='gray')
plt.axis('off')
plt.show()
````

````python
# 2x2のカラー画像（RGB）を作成
color_image = [[[255, 0, 0], [0, 255, 0]],
               [[0, 0, 255], [255, 255, 0]]]
#color_image = np.array([[[255, 0, 0], [0, 255, 0]],
#               [[0, 0, 255], [255, 255, 0]]])
# [R,G,B]: [255,0,0] は赤、[0,255,0]は緑、[0,0,255]は青、[255,255,0]は黄
# [[赤,緑],[青,黄]] のように色情報を要素に持つ2x2の画像になっている　(2x2x3：3次元テンソル）
# カラー画像を表示
plt.imshow(color_image)
plt.axis('off')
plt.show()
````

````python
# グレースケール
gradient_image = np.linspace(0, 255, 10, dtype=np.uint8).reshape(1, 10) # ベクトルを1x10の行列に変換
# dtype=np.uint8: 0〜255の整数にする（画像の画素値によく使われる型）
# linspace: linear space 区間を等間隔に分割した等差数列を返す
# linspace(start, stop, num)
print(gradient_image)
plt.imshow(gradient_image, 'gray')
plt.axis('off')
plt.show()
````

## サンプル画像と3次元データ*

scikit-image (`skimage`) には、医療・生物の画像を含むサンプル画像が入っている。`data.関数名()` でnumpy配列として取り出せる（一部は初回にダウンロードするので時間がかかる）。
サンプル画像の一覧は[ドキュメント](https://scikit-image.org/docs/stable/api/skimage.data.html)にある。

````python
from skimage import data

# 医療・生物に関係するサンプル画像
images = {
    "retina": data.retina(),                              # 眼底写真
    "immunohistochemistry": data.immunohistochemistry(),  # 免疫組織化学染色
    "cell": data.cell(),                                  # 細胞の顕微鏡画像
    "human_mitosis": data.human_mitosis(),                # 細胞分裂中の核の蛍光画像
}

for name in images:
    img = images[name]
    print(name, img.shape)  # カラー画像は (縦, 横, 3)、モノクロ画像は (縦, 横)
    plt.imshow(img, cmap="gray")  # cmap はモノクロ画像のときだけ使われる
    plt.title(name)
    plt.axis("off")
    plt.show()
````

`data.cells3d()` は細胞の核と細胞膜を蛍光顕微鏡で深さを変えながら60枚撮影した3次元データ。
形は (深さz, チャンネル, 縦, 横) = (60, 2, 256, 256) の4次元配列。

CTやMRIも同じように3次元データであり、スライス（断面）を切り出して2次元の画像として見る。

次のコードは、実行して結果を見るだけでよい。

````python
c = data.cells3d()[:, 1]   # チャンネル1 = 核の染色
print(c.shape)             # (深さ, 縦, 横) = (60, 256, 256)

# 深さ z=30 の断面（上から見た画像）
plt.imshow(c[30], cmap="gray")
plt.title("c[30]")
plt.axis("off")
plt.show()

# 縦の位置 128 で切った断面（横から見た画像）
plt.imshow(c[:, 128, :], cmap="gray")
plt.title("c[:, 128, :]")
plt.axis("off")
plt.show()

# 深さ方向に、各画素の最大値をとった画像（最大値投影）
plt.imshow(c.max(axis=0), cmap="gray")
plt.title("c.max(axis=0)")
plt.axis("off")
plt.show()
````
