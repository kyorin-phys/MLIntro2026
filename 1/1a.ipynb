## 要素の取り出しと型変換

* **インデックス**：リスト・タプル・文字列の要素は`[番号]`で取り出す。番号は**0から始まる**（1番目の要素は`[0]`）。`[-1]`は最後の要素。
* **スライス**：`[開始:終了]`で範囲を取り出す。終了の番号は含まない（`range`と同じ）。
* リストの要素は`e[1] = 20`のように変更できるが、タプルの要素は変更できない。
* **アンパック**：`w, h = (1920, 1080)`のように、タプルの要素を複数の変数に一度に代入できる。
* **辞書**：`h["physics"]`のようにキーを指定して値を取り出す。
* **型変換**：`int("123")`（文字列→整数）、`float("3.5")`、`str(10)`（数値→文字列）
* **f文字列**：`f"..."`の中に`{変数}`と書くと、変数の値が埋め込まれる。

````python
e = [1, 2, 3, 4, 5]
print(e[0], e[1], e[-1]) # 1番目、2番目、最後の要素
print(e[1:3])            # 2番目から3番目まで（e[3] は含まない）
e[1] = 20                # 2番目の要素を変更
print(e, len(e), sum(e)) # 要素の数と合計

w, h = (1920, 1080)      # アンパック
print(w, h)

h = {"biology": 70, "chemistry": 68, "physics": 75}
print(h["physics"])      # 辞書はキーで取り出す

n = int("123")           # 文字列を整数に変換
name = "taro"
print(f"Hello, {name}! n+1={n+1}") # f文字列
````

## リストに対する演算*
* 要素の追加
    * append() 要素を末尾に追加
    * extend()　リストを末尾に追加
    * insert()　指定した位置に挿入

* 要素の削除
    * remove()　指定した要素のうち最初のものを削除
    * pop() 末尾の要素を削除
    * clear()　空にする
    
* 要素の検索とカウント
    * index()
    * count()
* ソートと逆順
    * sort()
    * reverse()

````python
# 要素の追加
a = [1, 2, 3]
print(a)
a.append(4) # 末尾に4を追加
print(a)
a.extend([5, 6]) # [5,6]というリストの要素を追加して拡張
print(a)
a.insert(1, 1.5) # 1番目に1.5を挿入
print(a)
````

````python
# 要素の削除
a = [1, 2, 3, 1]
print(a)
a.remove(1) # 最初に現れる1を削除 
print(a)
b = a.pop() # 末尾の要素を取り出す
print(b, a)
a.clear() # 要素をすべて削除
print(a)
````

````python
# 要素の検索とカウント
a = [1, 2, 2, 3, 3, 3]
print(a)
print(a.index(2)) # 要素2が最初に現れるインデックス
print(a.count(2)) # 要素2が何個あるか数える
````

````python
# ソートと逆順
a = [2, 1, 3]
print(a)
print(sorted(a)) # aそのものは変化しない
print(a)
a.sort() # aの要素を並び替え
print(a)
a.reverse() # 逆順に並び替え
print(a)
````

## リスト内包表記*

python独特の記法。ループより高速なのでよく使われる。

```python
[式 for 変数 in 繰り返し要素]
```
forループを1行に短縮したような形
```python
リスト名=[]
for 変数 in 繰り返し要素:
    リスト名.append(式)
```

````python
# 例１
numbers = [1, 2, 3, 4, 5]
doubled = [x*2 for x in numbers]
print(doubled)

# for 文で書くと
doubled = [] # 空のリストを作る
for x in numbers:
  doubled.append(x*2) # 要素を追加する
print(doubled)
````

````python
# 例2
# 3項演算子（条件式）と組み合わせる
numbers = [1, 2, 3, 4, 5]
# 偶数ならそのまま、奇数なら0にする
result = [x if x % 2 == 0 else 0 for x in numbers]
print(result)  # 出力: [0, 2, 0, 4, 0]

# 同じことをfor文で書くと
result = []
for x in numbers:
  if x % 2 == 0:
    result.append(x)
  else:
    result.append(0)
print(result)
# 条件が3つ以上になる場合は、for文で書いた方が読みやすい。
````

## 3項演算子*
```python
真のときの値 if 条件 else 偽のときの値
```
リスト内包表記、3項演算子は簡潔に書けるが、慣れるまでは`for`文や`if`文が使えればよい。 

````python
# 例
a = 10
b = 20
max_value = a if a > b else b # a > b のときは a, それ以外のときは b を返す
print(a, b, max_value)

# if 文で書くと
a = 20
b = 10
if a > b:
  max_value = a
else:
  max_value = b
print(a, b, max_value)
# 4行分を1行に圧縮
````

## 数値の丸め*

Pythonの`round()`は**高校までの四捨五入と結果が違うことがある**ので注意。

| 関数 | 動作 | 2.5 | 3.5 | -2.5 |
|---|---|---|---|---|
| 高校の四捨五入 | 5以上は切り上げ | 3 | 4 | -3 |
| `round()` | 偶数への丸め | **2** | 4 | **-2** |
| `numpy.round()` | 偶数への丸め | **2.0** | 4.0 | **-2.0** |
| `math.floor()` | 切り捨て（小さい方の整数） | 2 | 3 | -3 |
| `math.ceil()` | 切り上げ（大きい方の整数） | 3 | 4 | -2 |

### なぜ違うのか
1. **偶数への丸め（Round Half to Even）**：ちょうど真ん中（…5）のときだけ、近い**偶数**の方に丸める。
   いつも切り上げると、大量のデータを丸めて合計したときに少しずつ大きい方へずれていく。偶数への丸めは切り上げと切り捨てが半々になるので、この偏りが生じない。統計や科学計算の標準的な方法（JIS Z 8401 でも採用）。
2. **小数は2進数で正確に表せない**：`2.675` はコンピュータの中では `2.67499999...` なので、`round(2.675, 2)` は 2.68 ではなく **2.67** になる（「真ん中」ではないので、偶数への丸め以前の問題）。

### 他のソフトでは
Excelの`ROUND()`、Cの`round()`、Fortranの`NINT()`は高校と同じ四捨五入（-2.5は-3）。
一方、Excel VBAの`Round()`、Rの`round()`はPythonと同じ偶数への丸め。
**ExcelとPythonで結果が1ずれたら、丸め方の違いを疑う。**

### どうすればよいか
* データ解析・機械学習では`round()`のままでよい（差はごくわずかで、むしろ偏りがない）。
* 成績や金額など、**高校の四捨五入どおりの結果が必要なとき**は`decimal`ライブラリを使う（下の`shishagonyu()`関数）。

````python
# round() の結果を予想してから実行してみよう
print(round(0.5), round(1.5), round(2.5), round(3.5)) # 偶数への丸め
print(round(2.675, 2)) # 小数第2位に丸める。2進数の誤差で 2.67 になる
````

````python
# 高校の四捨五入どおりに丸める関数
from decimal import Decimal, ROUND_HALF_UP

def shishagonyu(x, ndigits=0):
    # str(x) で「見たままの数字」にしてから丸めるので 2進数の誤差の影響を受けない
    q = Decimal(1).scaleb(-ndigits) # ndigits=0 なら 1, ndigits=2 なら 0.01
    r = Decimal(str(x)).quantize(q, rounding=ROUND_HALF_UP)
    return int(r) if ndigits == 0 else float(r)

print(shishagonyu(2.5), shishagonyu(3.5), shishagonyu(-2.5))
print(shishagonyu(2.675, 2))
````

### 小数の誤差*
0.1 のような小数も2進数では正確に表せないので、計算結果にごくわずかな誤差が出る。
小数どうしを`==`で比べるのは避け、`math.isclose()`を使う。

````python
import math

print(0.1 + 0.1 + 0.1 - 0.3) # = 0 ???
print(1.30 + 1.20) # = 2.50 ???
print(0.1 + 0.2 == 0.3, math.isclose(0.1 + 0.2, 0.3))
````

## 辞書の要素を調べる

```
d = {'A':'mitaka', 'B': 'musashino', 'C': 'chofu'}
```
という辞書があるとき、全ての要素にアクセスするには？

````python
d = {'A':'mitaka', 'B': 'musashino', 'C': 'chofu'}
print('keys:', d.keys())
print('values:', d.values())
print('items:', d.items())

for key in d.keys():
    print(key, d[key])
````

````python
for key, value in d.items():
    print(key, value)
````
